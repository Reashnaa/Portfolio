```python
# ================
# PROBLEM STATEMENT - Data Exploration
# ================
import pandas as pd
import datetime

pd.set_option('display.max_rows', None)        # Show all rows
pd.set_option('display.max_columns', None)     # Show all columns
pd.set_option('display.width', None)           # Use full width of display
pd.set_option('display.max_colwidth', None)    # Show full cell content

# Read file
df = pd.read_excel("resale-flat-prices-from-jan-2017.xlsx")
# print(df.tail(3))

# Check empty data
print("Empty data:")
print(df.isnull().sum())

print("-"*50)
 
# Check duplicate data - 210 rows
dup = df.duplicated().sum()
print("Duplicate data:", dup)

print("-"*50)

# Check wrong data
# ----- WRONG DATA 1 ----- 
# 993 and 1102 entered as floor area which is too large for HDB units. 
# The data entered may be mistaken as sqft data when the column expects sqm (for floor area)
# This means >= 300sqm is considered wrong data
floor_filter = df['floor_area_sqm'] >= 300
count1 = (floor_filter).sum()
print("Wrong Data (>= 300sqm floor area):", count1)

print("-"*50)

# ----- WRONG DATA 2 -----
# Date that is the future
# 2995 (lease commence date)
today = datetime.date.today()
lease_date_filter = df['lease_commence_date'] >= today.year
count2 = (lease_date_filter).sum()
print("Wrong Data (Future Lease Date):", count2)
```

```python
# Problem Scenario Exercise - Slide #13
# Chart shown is for the input 2 ROOM, BEDOK

# Import
import pandas as pd
import datetime

# Read file
df = pd.read_excel("resale-flat-prices-from-jan-2017.xlsx")
print("Original (count):")
print(df.count())
print("-"*50)

# ************************
# Part 1 - Data Wrangling
# ************************

# -----------------------
# Clean duplicate data
# -----------------------
df.drop_duplicates(inplace=True)

print("After Remove Duplicate (count):")
print(df.count())
print("-"*50)

# ------------------------------------------
# Clean wrong data set 1 - Wrong Floor Area
# ------------------------------------------
# Given size of dataset, filter and drop rows where the data is wrong
# Alternatively, one could replace with suitable values albeit not shown here. 
# If students choose to do so, enquire of their reasoning
floor_filter = df['floor_area_sqm'] >= 300
df.drop(df[floor_filter].index, inplace=True)

# --------------------------------------------------
# Clean wrong data set 2 - Wrong Lease Commence Date
# --------------------------------------------------
# Get today's date from the system
today = datetime.date.today()
#print(today.year)

# Remove rows where the 'lease_commence_date' is in the future 
# (i.e., lease year is greater than or equal to the current year).
# We identify those rows using boolean filtering, then drop them by index.
lease_date_filter = df['lease_commence_date'] >= today.year
df.drop(df[lease_date_filter].index, inplace=True)

print("After Fixed Wrong Data (count):")
print(df.count())
print("-"*50)

# -----------------
# Clean Empty Data
# -----------------
df.dropna(inplace=True)     # Drop using the existing data frame. 'inplace=True'

print("After Handle Missing Data (count):")
print(df.count())
print("-"*50)
```

```python
# ****************
# Part 2 - Program
# ****************

# Display and choose flat types
# Extract all the unique values from the 'flat_type' column.
flat_type = df['flat_type'].unique()
print("Flat types")
# print(type(df))

# Loop through the flat_type to display list of flats
for i in flat_type: 
    print(i)

# Prompt user for the flat type
flat = input("Choose the flat type:")

# Select data set based on the flat type input by user
flat_filter = df['flat_type'] == flat
df1 = df.loc[flat_filter, ["month", "town", "resale_price"]]

# Extract all the unique values from the 'town' column.
town = df['town'].unique()

# Display the town list using for loop
print("Town")
for i in town: 
    print(i)

print("-"*50)

# Prompt user for the town input
tw = input("Choose the town:")

# Select data set based on the town input by user
tw_filter = df1["town"] == tw
df2 = df1.loc[tw_filter, ['month', 'resale_price']]

# Group by month column and find the mean
df3 = df2.groupby('month').mean()['resale_price']
print(df3)

df3.plot(
        title= "Resale Price",
        xlabel= 'Year-Month', ylabel= "Price $",
        kind='bar')
```
